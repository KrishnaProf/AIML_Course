```python
import sqlite3
import pandas as pd

# Step 1: Connect to SQLite database (file-based)
conn = sqlite3.connect("ecommerce.db")
cursor = conn.cursor()

# Step 2: Create tables with constraints
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL UNIQUE,
    city TEXT NOT NULL
);
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_id INTEGER NOT NULL,
    amount REAL NOT NULL,
    order_date DATE NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
);
""")

# Step 3: Insert sample data using executemany()
customers_data = [
    ("Alice", "New York"),
    ("Bob", "Chicago"),
    ("Charlie", "New York")
]

orders_data = [
    (1, 120.50, "2026-05-01"),
    (2, 75.00, "2026-05-02"),
    (3, 200.00, "2026-05-03"),
    (1, 90.00, "2026-05-04")
]

cursor.executemany("INSERT OR IGNORE INTO customers (name, city) VALUES (?, ?);", customers_data)
cursor.executemany("INSERT INTO orders (customer_id, amount, order_date) VALUES (?, ?, ?);", orders_data)

conn.commit()

# Step 4: Query customers from New York with orders > 100
query = """
SELECT c.name, c.city, o.amount
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
WHERE c.city = 'New York' AND o.amount > 100;
"""

cursor.execute(query)
results = cursor.fetchall()

# Step 5: Save results to Pandas DataFrame
df = pd.DataFrame(results, columns=["Customer Name", "City", "Order Amount"])

# Step 6: Output results
print(df)

# Step 7: Close connection
conn.close()
```
